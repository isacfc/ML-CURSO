```python
df_asia = df[df["origin"] == "Asia"]
df_asia
df_asia = df_asia[["vehicle_weight","model_year"]]

df_asia = df_asia.head(7)
```

```python
X = df_asia.values
X
```

```python
X2=X.T
X2
```

```python
XTX = X2@X
XTX
XTX.shape
```

```python
XTX = inv(XTX)
```

```python
y = np.array([1100,1300,800,900,1000,1100,1200])
w = XTX @ X2 @ y
w= w.sum()
w
```
